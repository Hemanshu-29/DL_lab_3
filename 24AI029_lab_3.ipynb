```python
# =====================================================================
# MASTER ASSIGNMENT: CIFAR-100 REGULARIZATION & OPTIMIZERS STUDY
# INCLUDES: HIGH-SPEED DIRECT GOOGLE CLOUD DOWNLOAD BYPASS
# =====================================================================

import os
import pickle
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt

# ---------------------------------------------------------------------
# STEP 1: HIGH-SPEED DIRECT DATASET DOWNLOAD & INITIALIZATION
# ---------------------------------------------------------------------
print("--- [Step 1] Initializing High-Speed Server-to-Server Download ---")

# 1. Create the exact destination directory path that TensorFlow Keras expects
!mkdir -p /root/.keras/datasets/

# 2. Download directly from the official University of Toronto server using Google's backbone network
# This bypasses your local network speed entirely and completes in 2-3 seconds.
!wget -q --show-progress -O /root/.keras/datasets/cifar-100-python.tar.gz https://toronto.edu

print("File successfully placed in system cache folder.")

# 3. Load and unpack the data locally from the downloaded archive
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar100.load_data()

# 4. Normalize image pixel values from [0, 255] to [0.0, 1.0] for stable gradient descent
x_train, x_test = x_train / 255.0, x_test / 255.0

print(f"CIFAR-100 loaded! Train size: {x_train.shape}, Test size: {x_test.shape}\n")


# ---------------------------------------------------------------------
# UTILITY: REUSABLE GRAPH PLOTTING FUNCTION FOR REGULARIZATION
# ---------------------------------------------------------------------
def plot_individual_experiment(history, title):
    """Generates isolated detailed analysis plots for individual regularization tests."""
    plt.figure(figsize=(14, 4))

    # Loss Plot
    plt.subplot(1, 2, 1)
    plt.plot(history.history['loss'], label='Train Loss', color='royalblue', lw=2)
    plt.plot(history.history['val_loss'], label='Val Loss', color='crimson', lw=2)
    plt.title(f'{title} - Loss Curves')
    plt.xlabel('Epochs')
    plt.ylabel('Loss Value')
    plt.legend()
    plt.grid(True, linestyle='--')

    # Accuracy Plot
    plt.subplot(1, 2, 2)
    plt.plot(history.history['accuracy'], label='Train Acc', color='royalblue', lw=2)
    plt.plot(history.history['val_accuracy'], label='Val Acc', color='crimson', lw=2)
    plt.title(f'{title} - Accuracy Curves')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy Rate')
    plt.legend()
    plt.grid(True, linestyle='--')

    plt.tight_layout()
    plt.show()


# =====================================================================
# PART 1: REGULARIZATION ISOLATION EXPERIMENTS
# =====================================================================
print("\n=== [PART 1] RUNNING REGULARIZATION EXPERIMENTS ===")

# --- Experiment 1A: Baseline Model (No Regularization) ---
print("\nRunning Experiment 1A: Baseline (No Regularization)...")
model_baseline = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(32, 32, 3)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.Dense(100, activation='softmax')
])
model_baseline.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history_baseline = model_baseline.fit(x_train, y_train, epochs=20, batch_size=128, validation_data=(x_test, y_test), verbose=1)
plot_individual_experiment(history_baseline, "1A. Baseline (No Regularization)")


# --- Experiment 1B: Dropout Regularization ---
print("\nRunning Experiment 1B: Dropout Regularization...")
model_dropout = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(32, 32, 3)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25), # Convolutional layer structural dropout

    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),

    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.5),  # Dense bottleneck layer layer dropout
    layers.Dense(100, activation='softmax')
])
model_dropout.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history_dropout = model_dropout.fit(x_train, y_train, epochs=20, batch_size=128, validation_data=(x_test, y_test), verbose=1)
plot_individual_experiment(history_dropout, "1B. Dropout (0.25 & 0.5)")


# --- Experiment 1C: Early Stopping Control ---
print("\nRunning Experiment 1C: Early Stopping...")
model_es = tf.keras.models.clone_model(model_baseline)
model_es.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Set patience to 3 epochs; if val_loss fails to improve, execution halts and reverts to best weights
es_callback = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)
history_es = model_es.fit(x_train, y_train, epochs=25, batch_size=128, validation_data=(x_test, y_test), callbacks=[es_callback], verbose=1)
plot_individual_experiment(history_es, "1C. Early Stopping Control")


# --- Experiment 1D: L2 Weight Regularization (Weight Decay) ---
print("\nRunning Experiment 1D: L2 Weight Penalties...")
model_l2 = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', padding='same', kernel_regularizer=regularizers.l2(0.001), input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu', kernel_regularizer=regularizers.l2(0.001)),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(512, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
    layers.Dense(100, activation='softmax')
])
model_l2.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history_l2 = model_l2.fit(x_train, y_train, epochs=20, batch_size=128, validation_data=(x_test, y_test), verbose=1)
plot_individual_experiment(history_l2, "1D. L2 Weight Decay")


# =====================================================================
# PART 2: OPTIMIZER COMPARISON STUDY
# =====================================================================
print("\n=== [PART 2] RUNNING OPTIMIZER STUDY ===")

# Standardized architecture function to ensure comparison integrity across optimizers
def make_neutral_architecture():
    return models.Sequential([
        layers.Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(32, 32, 3)),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
        layers.MaxPooling2D((2, 2)),
        layers.Flatten(),
        layers.Dense(256, activation='relu'),
        layers.Dense(100, activation='softmax')
    ])

# Dictionary housing the optimizers to be tested sequentially
optimizer_suite = {
    'SGD (with Momentum)': tf.keras.optimizers.SGD(learning_rate=0.01, momentum=0.9),
    'Adam': tf.keras.optimizers.Adam(learning_rate=0.001),
    'AdaGrad': tf.keras.optimizers.Adagrad(learning_rate=0.01),
    'RMSprop': tf.keras.optimizers.RMSprop(learning_rate=0.001)
}

opt_histories = {}

# Execute back-to-back training metrics calculations
for title, optimizer_obj in optimizer_suite.items():
    print(f"\nTraining standard network variant utilizing: {title}...")
    opt_model = make_neutral_architecture()
    opt_model.compile(optimizer=optimizer_obj, loss='sparse_categorical_crossentropy', metrics=['accuracy'])

    run_log = opt_model.fit(x_train, y_train, epochs=12, batch_size=128, validation_data=(x_test, y_test), verbose=1)
    opt_histories[title] = run_log.history

# --- Generating Combined Comparison Plots ---
plt.figure(figsize=(16, 6))

# Plot 2A: Consolidated Multi-Optimizer Loss Profile
plt.subplot(1, 2, 1)
for title in opt_histories:
    plt.plot(opt_histories[title]['val_loss'], label=f"{title} [Val]", linestyle='--')
    plt.plot(opt_histories[title]['loss'], label=f"{title} [Train]", alpha=0.3)
plt.title('Optimizer Loss Performance Profiles')
plt.xlabel('Epochs')
plt.ylabel('Loss Value')
plt.legend()
plt.grid(True)

# Plot 2B: Consolidated Multi-Optimizer Validation Accuracy Performance Tracking
plt.subplot(1, 2, 2)
for title in opt_histories:
    plt.plot(opt_histories[title]['val_accuracy'], label=f"{title} Accuracy Score", linewidth=2.5)
plt.title('Optimizer Validation Accuracy Trajectories')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# --- Print Final Numerical Statistical Performance Table ---
print("\n" + "="*60)
print(f"{'Optimizer Strategy':<25} | {'Final Train Acc':<15} | {'Final Val Acc':<13}")
print("="*60)
for title in opt_histories:
    train_score = opt_histories[title]['accuracy'][-1] * 100
    val_score = opt_histories[title]['val_accuracy'][-1] * 100
    print(f"{title:<25} | {train_score:>13.2f}% | {val_score:>11.2f}%")
print("="*60)
```
